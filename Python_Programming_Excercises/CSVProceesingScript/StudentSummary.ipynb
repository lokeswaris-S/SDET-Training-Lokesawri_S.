```python
import csv

```

```python
def read_csv(input_file):
    students = []

    try:
        with open(input_file, "r") as file:
            reader = csv.DictReader(file)

            for row in reader:
                try:
                    row["Marks"] = int(row["Marks"])
                    students.append(row)
                except:
                    print("Skipping invalid marks:", row["Marks"])

    except FileNotFoundError:
        print("File not found")
    except:
        print("Invalid CSV format")

    return students
```

```python
def calculate_statistics(data):
  
    if not data:
        print("No valid data found.")
        return None

    total_students = len(data)

    marks_list = [student["Marks"] for student in data]

    average_marks = sum(marks_list) / total_students
    highest_marks = max(marks_list)

    toppers = [student["Name"] for student in data if student["Marks"] == highest_marks]

    summary = {
        "Total Students": total_students,
        "Average Marks": round(average_marks, 2),
        "Highest Marks": highest_marks,
        "Topper(s)": ", ".join(toppers)
    }

    return summary
```

```python
def write_summary(output_file, summary):
    try:
        with open(output_file, "w", newline="") as file:
            writer = csv.writer(file)

            writer.writerow(summary.keys())
            writer.writerow(summary.values())

    except:
        print("Error writing output file")
```

```python
def write_summary(output_file, summary):

    try:
        with open(output_file, "w", newline="") as file:
            writer = csv.writer(file)

          
            writer.writerow(["Average Marks", "Highest Marks", "Topper(s)"])

         
            writer.writerow([
                summary["Average Marks"],
                summary["Highest Marks"],
                summary["Topper(s)"]
            ])

    except Exception as e:
        print("Error while writing summary file:", e)
```

```python
input_file = "students.csv"
output_file = "summary.csv"

student_data = read_csv(input_file)


summary = calculate_statistics(student_data)

if summary:
    write_summary(output_file, summary)

    print("\nSummary Generated Successfully:")
    print(summary)
```
